```python
import numpy as np
from scipy.ndimage import center_of_mass
from scipy.spatial import cKDTree
```

```python
path = "/zjbd/zd1/shechtmanlab/omer/MS2/outputs/020626/STAGE-11/New-02-v3-ST11-12/New-02-v3-ST11-12/masks/z_stack_t0_seg_masks.npz"
```

```python
mask = np.load(path)["masks"]
labels = np.unique(mask); labels = labels[labels > 0]
coms = np.array([com[1:] for com in center_of_mass(mask, mask, labels)])  # Y, X only
tree = cKDTree(coms)
dists, _ = tree.query(coms, k=2)          # k=2: skip self (dist=0)
print(f"Median nearest-neighbour distance: {np.median(dists[:, 1]):.1f} px")
print(f"Suggested search_range: {np.median(dists[:, 1]) * 0.5:.1f} px")
```
